# Section D — K-th smallest pairwise sum

Find the 1-based k-th smallest sum among all pairs of distinct indices `i < j`. The optimized method sorts once, then binary-searches the possible sum value.

```python
from time import perf_counter

def kth_sum_naive(values, k):
    sums = [values[i] + values[j] for i in range(len(values)) for j in range(i + 1, len(values))]
    if not 1 <= k <= len(sums):
        raise ValueError("k is outside the valid range")
    sums.sort()
    return sums[k - 1]

def count_at_most(values, limit):
    count = 0
    right = len(values) - 1
    for left in range(len(values)):
        while left < right and values[left] + values[right] > limit:
            right -= 1
        count += max(0, right - left)
    return count

def kth_sum_optimized(values, k):
    if len(values) < 2:
        raise ValueError("At least two values are needed")
    total_pairs = len(values) * (len(values) - 1) // 2
    if not 1 <= k <= total_pairs:
        raise ValueError("k is outside the valid range")
    arr = sorted(values)
    low, high = arr[0] + arr[1], arr[-2] + arr[-1]
    while low < high:
        mid = (low + high) // 2
        if count_at_most(arr, mid) >= k:
            high = mid
        else:
            low = mid + 1
    return low
```

```python
numbers = [1, 2, 3, 4]
k = 3
for method in (kth_sum_naive, kth_sum_optimized):
    start = perf_counter()
    answer = method(numbers, k)
    elapsed = perf_counter() - start
    print(f"{method.__name__}: {answer}, {elapsed:.8f}s")

assert kth_sum_naive(numbers, k) == kth_sum_optimized(numbers, k)
```
