# Section B — Sum of all primes below N

Compare naive trial division, square-root trial division, and the Sieve of Eratosthenes. The limit is exclusive.

```python
from math import isqrt
from time import perf_counter

def prime_sum_naive(n):
    total = 0
    for x in range(2, n):
        prime = True
        for d in range(2, x):
            if x % d == 0:
                prime = False
                break
        if prime:
            total += x
    return total

def prime_sum_sqrt(n):
    total = 0
    for x in range(2, n):
        if x == 2:
            total += 2
        elif x % 2:
            for d in range(3, isqrt(x) + 1, 2):
                if x % d == 0:
                    break
            else:
                total += x
    return total

def prime_sum_sieve(n):
    if n <= 2:
        return 0
    prime = [True] * n
    prime[0:2] = [False, False]
    for p in range(2, isqrt(n - 1) + 1):
        if prime[p]:
            prime[p*p:n:p] = [False] * len(range(p*p, n, p))
    return sum(i for i, flag in enumerate(prime) if flag)
```

```python
N = 1000  # Change this value to try another input
methods = [prime_sum_naive, prime_sum_sqrt, prime_sum_sieve]
results = []
for method in methods:
    start = perf_counter()
    answer = method(N)
    elapsed = perf_counter() - start
    results.append(answer)
    print(f"{method.__name__}: sum={answer}, time={elapsed:.8f}s")

assert len(set(results)) == 1, "The implementations should agree."
print("All methods agree.")
```
