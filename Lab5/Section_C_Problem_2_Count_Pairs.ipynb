# Section C — Count pairs with a given sum

Count index pairs `(i, j)` with `i < j` whose values add to the target. Repeated values at different positions count as different pairs.

```python
from time import perf_counter

def count_pairs_naive(values, target):
    count = 0
    for i in range(len(values)):
        for j in range(i + 1, len(values)):
            if values[i] + values[j] == target:
                count += 1
    return count

def count_pairs_hashmap(values, target):
    seen = {}
    count = 0
    for value in values:
        count += seen.get(target - value, 0)
        seen[value] = seen.get(value, 0) + 1
    return count
```

```python
numbers = [1, 5, 7, -1, 5]
target = 6
for method in (count_pairs_naive, count_pairs_hashmap):
    start = perf_counter()
    result = method(numbers, target)
    elapsed = perf_counter() - start
    print(f"{method.__name__}: {result} pair(s), {elapsed:.8f}s")

assert count_pairs_naive(numbers, target) == count_pairs_hashmap(numbers, target)
```
